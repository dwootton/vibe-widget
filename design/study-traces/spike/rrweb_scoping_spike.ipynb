```python
import pandas as pd
from spike import Spike, Replay, stats
pd.DataFrame({"patient": ["SECRET_PATIENT_4471", "SECRET_PATIENT_9902"], "dx": ["T2D", "CKD"]})
```

```python
# A: rrweb on the whole notebook page
a = Spike(mode="document"); a
```

```python
# B: whole page, but block every other cell
b = Spike(mode="document", block_selector=".jp-Cell:not(:has(.vw-spike-root)), .jp-SideBar, .jp-DirListing"); b
```

```python
# C: widget mounted in a same-origin iframe, rrweb only inside it
c = Spike(mode="iframe"); c
```

```python
pd.DataFrame({name: stats(w.events) for name, w in [("A document", a), ("B document+block", b), ("C iframe", c)]})
```

```python
Replay(events=c.events)
```
